```python
import polars as pl
from IPython.display import display
from relisten import config
from relisten.official_models import build_official_features
from relisten.evaluation import evaluate_features
```

```python
features = build_official_features("iALS")
display(features.select("model_end_day").group_by("model_end_day").len())
```

```python
models = {f"iALS_seed{seed}":f"iALS_seed{seed}" for seed in (41,42,43)}
metrics = evaluate_features(features,models,"ials_model_metrics.csv")
display(metrics.filter((pl.col("route")=="C") & (pl.col("split")=="test") & (pl.col("lower_days")==config.LOWER_BOUND_DAYS)))
```
